```python
def removeDuplicates(s):
    result = ""
    for char in s:
        if char not in result:
            result += char
    return result

print(removeDuplicates("programming"))
```

```python
def sum(numbers):
  total=0
  for num in numbers:
    total+=num
  return total
numbers=[1,2,3,4,5]
print(sum(numbers))
```

```python
def remove(list1):
  new=[]
  for i in list1:
    if i not in new:
      new.append(i)
  return new
list1=[1,2,3,4,4,5,6]
```

```python
s = "abcde"
is_unique = True
for i in range(len(s)):
    for j in range(i + 1, len(s)):
        if s[i] == s[j]:
            is_unique = False
            break
    if not is_unique:
        break
is_unique
```
